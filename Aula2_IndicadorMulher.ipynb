O `.xls` do IBGE mistura **título**, cabeçalho em várias linhas e **notas no rodapé**.  
Cortamos isso e deixamos só: **nomes das colunas + números**.

Tabela 1.1.1 - Número médio de horas semanais dedicadas aos cuidados de pessoas e/ou afazeres domésticos das pessoas de 14 anos ou mais de idade, na semana de referência, por sexo e cor ou raça, segundo Grandes Regiões e Unidades da Federação - Brasil - 2022	

```python
# Lê sem usar a 1ª linha como nome de coluna (senão o título vira "coluna")
bruto = pd.read_excel(path, engine="xlrd", header=None)

# Linhas 0–7: título/cabeçalho  |  8–40: resultados  |  41+: fonte e notas
indicador_1 = bruto.iloc[8:41].copy()
indicador_1.columns = [
    "uf_regiao",
    "total",
    "total_branca",
    "total_preta_parda",
    "homem_branca",
    "homem_preta_parda",
    "mulher_branca",
    "mulher_preta_parda",
]
indicador_1 = indicador_1.reset_index(drop=True)

# horas são número, não texto
for col in indicador_1.columns[1:]:
    indicador_1[col] = pd.to_numeric(indicador_1[col], errors="coerce")

indicador_1
```

```python
# Média nacional = linha Brasil (dado do IBGE), não a média das linhas da tabela
media_nacional = indicador_1.loc[indicador_1["uf_regiao"] == "Brasil", "total"].item()
media_nacional
```

```python
media_nacional
```

Qual(is) estado(s) está acima da média nacional de afazeres domésticos?

```python
regioes = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]

estados = indicador_1[~indicador_1["uf_regiao"].isin(["Brasil"] + regioes)]

acima_media = estados.loc[
    estados["total"] > media_nacional,
    ["uf_regiao", "total"],
].sort_values("total", ascending=False)

print(f"Média nacional (Brasil): {media_nacional:.2f} horas/semana")
print(f"Estados acima da média: {len(acima_media)}")
acima_media
```

Quais estados estão abaixo da média?

```python
regioes = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]

estados = indicador_1[~indicador_1["uf_regiao"].isin(["Brasil"] + regioes)]

abaixo_media = estados.loc[
    estados["total"] < media_nacional,
    ["uf_regiao", "total"],
].sort_values("total", ascending=False)

print(f"Média nacional (Brasil): {media_nacional:.2f} horas/semana")
print(f"Estados abaixo da média: {len(abaixo_media)}")
abaixo_media
```

Qual é o percentual de diferença no Brasil que mulheres trabalham mais que homens em casa?

```python

brasil = indicador_1.loc[indicador_1["uf_regiao"] == "Brasil"].iloc[0]

homem = (brasil["homem_branca"] + brasil["homem_preta_parda"]) / 2
mulher = (brasil["mulher_branca"] + brasil["mulher_preta_parda"]) / 2

# quanto % a mais as mulheres trabalham em relação aos homens
percentual = (mulher - homem) / homem * 100

print(f"Homens (média das cores):   {homem:.2f} h/semana")
print(f"Mulheres (média das cores): {mulher:.2f} h/semana")
print(f"Comparação das médias simples por cor ou raça: mulheres +{percentual:.1f}%.")
print("Essa média simples não representa o percentual nacional ponderado por população.")

# recorte por cor (a tabela traz isso de fato)
perc_branca = (brasil["mulher_branca"] / brasil["homem_branca"] - 1) * 100
perc_preta_parda = (brasil["mulher_preta_parda"] / brasil["homem_preta_parda"] - 1) * 100
print(f"Branca:        mulheres +{perc_branca:.1f}%")
print(f"Preta ou parda: mulheres +{perc_preta_parda:.1f}%")
```

## Resolução dos exercícios

A tabela reúne **pessoas pretas ou pardas** na mesma categoria; não permite analisar somente pessoas pretas. Os cálculos abaixo usam esse agrupamento.

**Diferença percentual = (horas do grupo comparado − horas do grupo de referência) / horas do grupo de referência × 100.**

Resultado positivo indica mais horas; negativo indica menos horas. Os cálculos usam os valores originais, com arredondamento apenas na exibição.


### 1. Minas Gerais — mulheres pretas ou pardas em relação às mulheres brancas

```python
dados = indicador_1.loc[indicador_1["uf_regiao"] == "Minas Gerais"].iloc[0]

mulheres_pretas_pardas = dados["mulher_preta_parda"]
referencia = dados["mulher_branca"]
percentual = (mulheres_pretas_pardas - referencia) / referencia * 100

print(f"Minas Gerais: {percentual:+.2f}%".replace(".", ","))
```

### 2. Brasil — mulheres pretas ou pardas em relação aos homens brancos

```python
dados = indicador_1.loc[indicador_1["uf_regiao"] == "Brasil"].iloc[0]

mulheres_pretas_pardas = dados["mulher_preta_parda"]
referencia = dados["homem_branca"]
percentual = (mulheres_pretas_pardas - referencia) / referencia * 100

print(f"Brasil: {percentual:+.2f}%".replace(".", ","))
```

### 3. Amapá — mulheres pretas ou pardas em relação às mulheres brancas

```python
dados = indicador_1.loc[indicador_1["uf_regiao"] == "Amapá"].iloc[0]

mulheres_pretas_pardas = dados["mulher_preta_parda"]
referencia = dados["mulher_branca"]
percentual = (mulheres_pretas_pardas - referencia) / referencia * 100

print(f"Amapá: {percentual:+.2f}%".replace(".", ","))
```

No Amapá, o sinal negativo significa que mulheres pretas ou pardas dedicam **menos horas**, em média, que mulheres brancas.

### 4. Rio Grande do Sul — mulheres pretas ou pardas em relação às mulheres brancas

```python
dados = indicador_1.loc[indicador_1["uf_regiao"] == "Rio Grande do Sul"].iloc[0]

mulheres_pretas_pardas = dados["mulher_preta_parda"]
referencia = dados["mulher_branca"]
percentual = (mulheres_pretas_pardas - referencia) / referencia * 100

print(f"Rio Grande do Sul: {percentual:+.2f}%".replace(".", ","))
```

### 5. Comparação por Grande Região

Usamos a linha de cada região fornecida pelo IBGE.

```python
regioes = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]

for regiao in regioes:
    dados = indicador_1.loc[indicador_1["uf_regiao"] == regiao].iloc[0]
    percentual = (dados["mulher_preta_parda"] - dados["mulher_branca"]) / dados["mulher_branca"] * 100
    print(f"{regiao}: {percentual:+.2f}%".replace(".", ","))
```

### 6. Comparação em cada estado do Sudeste

```python
estados_sudeste = ["Minas Gerais", "Espírito Santo", "Rio de Janeiro", "São Paulo"]

for estado in estados_sudeste:
    dados = indicador_1.loc[indicador_1["uf_regiao"] == estado].iloc[0]
    percentual = (dados["mulher_preta_parda"] - dados["mulher_branca"]) / dados["mulher_branca"] * 100
    print(f"{estado}: {percentual:+.2f}%".replace(".", ","))
```
