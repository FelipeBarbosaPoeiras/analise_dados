```python
import pandas as pd

ARQUIVO_ORIGINAL = "Tabela_3_Areas_gerais_de_formacao_na_graduacao__1_.xlsx"
```

```python
raw = pd.read_excel(ARQUIVO_ORIGINAL, header=None, skiprows=3)
raw = raw.iloc[:, 0:10]
raw.columns = [
    "Regiao",
    "Superior_Total", "Superior_Homens", "Superior_Mulheres",
    "CTEM_Total", "CTEM_Homens", "CTEM_Mulheres",
    "EducSaude_Total", "EducSaude_Homens", "EducSaude_Mulheres",
]

raw = raw.dropna(subset=["Regiao"])
raw = raw[~raw["Regiao"].astype(str).str.contains("Fonte", na=False)].reset_index(drop=True)
raw
```

```python
regioes_nomes = ["Brasil", "Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]
df_regioes = raw[raw["Regiao"].isin(regioes_nomes)].reset_index(drop=True)

for col in df_regioes.columns[1:]:
    df_regioes[col] = df_regioes[col].astype(int)

df_regioes
```

```python
df_superior = df_regioes[["Regiao", "Superior_Total", "Superior_Homens", "Superior_Mulheres"]].rename(
    columns={"Superior_Total": "Total", "Superior_Homens": "Homens", "Superior_Mulheres": "Mulheres"}
)

df_tecnologia = df_regioes[["Regiao", "CTEM_Total", "CTEM_Homens", "CTEM_Mulheres"]].rename(
    columns={"CTEM_Total": "Total", "CTEM_Homens": "Homens", "CTEM_Mulheres": "Mulheres"}
)

df_educacao_saude = df_regioes[["Regiao", "EducSaude_Total", "EducSaude_Homens", "EducSaude_Mulheres"]].rename(
    columns={"EducSaude_Total": "Total", "EducSaude_Homens": "Homens", "EducSaude_Mulheres": "Mulheres"}
)

df_superior
```
